```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
train=pd.read_csv("train.csv")
train.head()
```

```python
train.columns
```

```python
train.info()
```

```python
test=pd.read_csv("test.csv")
test.head()
```

```python
train.shape
```

```python
test.shape
```

```python
train=train.dropna()
train.info()
```

```python
test.shape
```

```python
test.info()
```

```python
X_train=train[["x"]]
Y_train=train["y"]
X_test=test[["x"]]
Y_test=test["y"]
```

```python
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.metrics import mean_absolute_error,r2_score
model=LinearRegression()
model.fit(X_train,Y_train)
pred=model.predict(X_test)
print("mean absolute error:",mean_absolute_error(Y_test,pred))

print("r2_score",r2_score(Y_test,pred))

```

```python
model.intercept_,model.coef_
```

```python
model.predict([[49]])
```

```python
#plotting the values on graph
plt.scatter(X_test,Y_test,label="realdata")
plt.plot(X_test,pred,color="red",label="model plane")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
rid=Ridge(alpha=10000.0)
rid.fit(X_train,Y_train)
predict_rid=rid.predict(X_test)
print("Mean_absolute_error:",mean_absolute_error(Y_test,predict_rid,))
print("r2_scor:",r2_score(Y_test,predict_rid))
```

```python
plt.scatter(X_test,Y_test,label="real data")
plt.plot(X_test,predict_rid,label="model_line",color="red")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
lasso=Lasso(alpha=10)
lasso.fit(X_train,Y_train)
predict_lasso=lasso.predict(X_test)
print("Mean_absolute_error:",mean_absolute_error(Y_test,predict_lasso))
print("r2_scor:",r2_score(Y_test,predict_lasso))
```

```python
plt.scatter(X_test,Y_test,label="real data")
plt.plot(X_test,predict_lasso,label="model_line",color="red")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
